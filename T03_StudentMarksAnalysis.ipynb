```python
# 3.count how many students scored 90+, 75–89, 50–74, below 50
marks = [78, 92, 45, 67, 88, 53, 99]
count_90_plus = 0
count_75_89 = 0
count_50_74 = 0
count_below_50 = 0
for i in marks:
    if i >= 90:
        count_90_plus = count_90_plus + 1
    elif i >= 75:
        count_75_89 = count_75_89 + 1
    elif i >= 50:
        count_50_74 = count_50_74 + 1
    else:
        count_below_50 = count_below_50 + 1

print("Number of students who scored 90+:", count_90_plus)
print("Number of students who scored 75–89:", count_75_89)
print("Number of students who scored 50–74:", count_50_74)
print("Number of students who scored below 50:", count_below_50)
```
